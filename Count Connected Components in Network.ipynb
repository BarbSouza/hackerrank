```python
#!/bin/python3

import math
import os
import random
import re
import sys
from collections import deque


#
# Complete the 'countIsolatedCommunicationGroups' function below.
#
# The function is expected to return an INTEGER.
# The function accepts following parameters:
#  1. 2D_INTEGER_ARRAY links
#  2. INTEGER n
#

def countIsolatedCommunicationGroups(links, n):
    count = 0
    seen = set()
    graph = {}
    for i in range(n):
        graph[i] = []
    for a, b in links:
        graph[a].append(b)
        graph[b].append(a)
    for computer in range(n):
        if computer not in seen:
            seen.add(computer)
            count += 1
            queue = deque()
            queue.append(computer)
            while queue:
                current = queue.popleft()
                for neighbour in graph[current]:
                    if neighbour not in seen:
                        seen.add(neighbour)
                        queue.append(neighbour)
                        
    return count
if __name__ == '__main__':
    links_rows = int(input().strip())
    links_columns = int(input().strip())

    links = []

    for _ in range(links_rows):
        links.append(list(map(int, input().rstrip().split())))

    n = int(input().strip())

    result = countIsolatedCommunicationGroups(links, n)

    print(result)
```
